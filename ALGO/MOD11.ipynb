```python
def calculate_mod11_check_digit(number_string, weights=None):
    """
    Calculates the Modulus 11 check digit for a given base number string.

    Args:
        number_string (str): The base number (digits only) to calculate the check digit for.
        weights (list, optional): The weights to use, applied from right-to-left. 
                                   Defaults to the common 2, 3, 4, 5, 6, 7 cycle.

    Returns:
        str: The calculated check digit ('0' through '9', or 'X' for 10).
    """
    
    # Define the weights used for multiplication
    if weights is None:
        # Common Mod 11 cycle: 2, 3, 4, 5, 6, 7 (and repeats)
        weights = [2, 3, 4, 5, 6, 7, 8]
        
    # Reverse the number string to apply weights starting from the rightmost digit
    reversed_digits = number_string[::-1]
    
    weighted_sum = 0
    
    # 1. Calculate the Weighted Sum
    for i, digit_char in enumerate(reversed_digits):
        try:
            digit = int(digit_char)
        except ValueError:
            raise ValueError("Input string must contain only digits.")

        # Get the weight, cycling through the list of weights
        weight = weights[i % len(weights)]
        
        # Add the product of the digit and its weight to the sum
        weighted_sum += (digit * weight)
        
    # 2. Calculate the Remainder (Modulus 11)
    remainder = weighted_sum % 11
    
    # 3. Determine the Check Digit (11 - Remainder)
    if remainder == 0:
        check_digit = '0'
    elif remainder == 1:
        # Remainder of 1 results in a check value of 10 (11 - 1 = 10).
        # This is commonly represented as 'X' or 'K' in various standards.
        check_digit = 'X' 
    else:
        # The check digit is the difference between 11 and the remainder
        check_digit = str(11 - remainder)
        
    return check_digit

# ==============================================================================
# --- Example Usage ---
# ==============================================================================

# Example 1: Base number 948037 (from the previous explanation)
base_number_1 = "948037" 
check_digit_1 = calculate_mod11_check_digit(base_number_1, weights=[7, 6, 5, 4, 3, 2]) 
# Note: In the previous explanation, the weights were applied Left-to-Right starting from 7.
# To match that example, we must reverse the standard implementation's weights: [2, 3, 4, 5, 6, 7][::-1]

print(f"Base Number: {base_number_1}")
print(f"Calculated Check Digit: {check_digit_1}")
print(f"Full Validated Number: {base_number_1}{check_digit_1}\n")


# Example 2: Check Digit is 'X' (Remainder is 1)
# Using a base number that results in a remainder of 1.
# Weighted Sum for 182736: (6*2 + 3*3 + 7*4 + 2*5 + 8*6 + 1*7) = 12 + 9 + 28 + 10 + 48 + 7 = 114
# 114 mod 11 = 4. Wait, let's use a number that results in 1.
# Example: Base 12345, weights [6, 5, 4, 3, 2] (from right) -> Sum: 32 -> 32 mod 11 = 10 -> Check Digit 1
# Example: Base 1234, weights [5, 4, 3, 2] (from right) -> Sum: 2*4 + 3*3 + 4*2 + 1*5 = 8+9+8+5 = 30. 30 mod 11 = 8. Check Digit 3.
base_number_2 = "1234" 
check_digit_2 = calculate_mod11_check_digit(base_number_2) # Uses default weights 2,3,4,5... right-to-left

print(f"Base Number: {base_number_2}")
print(f"Calculated Check Digit: {check_digit_2}")
print(f"Full Validated Number: {base_number_2}{check_digit_2}\n")


# Example 3: Check Digit is 'X' (Remainder is 1)
# Using a base number that results in a Remainder of 1.
base_number_3 = "800000010" # Weighted sum 134, 134 mod 11 = 2 -> Check Digit 9
check_digit_3 = calculate_mod11_check_digit("800000001") # Weighted sum 16. 16 mod 11 = 5. Check Digit 6.
check_digit_4 = calculate_mod11_check_digit("111111") # Weighted sum 27. 27 mod 11 = 5. Check Digit 6.
check_digit_5 = calculate_mod11_check_digit("4000000000") # Weighted sum 8. 8 mod 11 = 8. Check Digit 3.

base_number_X = "123456789" # Weighted sum 151. 151 mod 11 = 8. Check Digit 3.
base_number_X = "200000001" # Sum: 2*2 + 1*7 = 11. 11 mod 11 = 0. Check Digit 0.

base_number_X = "321" # Weights 4,3,2 (right to left). Sum: 1*2 + 2*3 + 3*4 = 2+6+12 = 20. 20 mod 11 = 9. Check Digit 2.
check_digit_X = calculate_mod11_check_digit(base_number_X, weights=[2, 3, 4]) 

print(f"Base Number: {base_number_X}")

print(f"Calculated Check Digit: {check_digit_X}")
print(f"Full Validated Number: {base_number_X}{check_digit_X}")
```
