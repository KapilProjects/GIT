```python
import hashlib

def hash_string_sha256(input_string):
    """
    Generates the SHA-256 hash of a given string.
    """
    # 1. Create a new SHA-256 hash object
    sha256 = hashlib.sha256()
    
    # 2. Update the hash object with the bytes of the string
    #    (Note: data must be encoded to bytes, e.g., 'utf-8')
    sha256.update(input_string.encode('utf-8'))
    
    # 3. Get the hexadecimal representation of the hash
    hex_digest = sha256.hexdigest()
    
    return hex_digest

# --- Example Usage ---
data = "Hello, world! This is a test for SHA256."
hash_result = hash_string_sha256(data)

print(f"Original String: '{data}'")
print(f"SHA-256 Hash: {hash_result}")
print(f"Hash Length: {len(hash_result)} characters (256 bits)")

# Changing even one character drastically changes the hash (Avalanche Effect)
data_modified = "Hello, world! This is a test for SHA255."
hash_result_modified = hash_string_sha256(data_modified)
print("\n--- Modified Data Hash ---")
print(f"Modified String: '{data_modified}'")
print(f"SHA-256 Hash: {hash_result_modified}")
```
